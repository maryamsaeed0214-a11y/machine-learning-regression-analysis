## **Task 3: Rename Columns**

**Instructions:**

1.  Create a dictionary to rename the following columns in your `data` DataFrame:
    *   `'InvoiceNo'` to `'transaction_id'`
    *   `'Description'` to `'item_name'`
    *   `'Quantity'` to `'order_quantity'`

## **Task 4: Data Selection and Filtering**

**Instructions:**

1.  Using `loc`, select and display the `item_name` and `item_price` columns for the first 10 rows of the `data` DataFrame.
2.  Filter the `data` DataFrame to show only orders where the `order_quantity` is greater than 1. Display the first 5 rows of this filtered DataFrame.
3.  Filter the `data` DataFrame to show orders where the `item_name` is 'WHITE HANGING HEART T-LIGHT HOLDER' **AND** the `order_quantity` is 6. Display all rows that meet this condition.
4.  Filter the `data` DataFrame to show orders where the `item_name` is 'WHITE METAL LANTERN' **OR** the `item_price` is less than 5. Display the first 5 rows of this filtered DataFrame.
5.  Using `iloc`, select and display all rows and only the first three columns of the `data` DataFrame.

```python
# item_name and item_price for first 10 rows
print(data.loc[:9, ["item_name", "item_price"]])

# order_quantity > 1
print(data[data["order_quantity"] > 1].head())

# item_name = WHITE HANGING HEART T-LIGHT HOLDER AND quantity = 6
print(data[
    (data["item_name"] == "WHITE HANGING HEART T-LIGHT HOLDER") &
    (data["order_quantity"] == 6)
])

# item_name = WHITE METAL LANTERN OR item_price < 5
print(data[
    (data["item_name"] == "WHITE METAL LANTERN") |
    (data["item_price"] < 5)
].head())

# First three columns using iloc
print(data.iloc[:, :3])
```

## **Task 5: Sorting and Unique Values Analysis**

**Instructions:**

1.  Sort the `data` DataFrame by the `item_price` column in ascending order. Display the first 5 rows of the sorted DataFrame.
2.  Sort the `data` DataFrame by the `order_quantity` column in descending order. Display the first 5 rows of the sorted DataFrame.
3.  Find and print the number of unique values in the `item_name` column.
4.  Display all the unique values (product names) present in the `item_name` column.
5.  Find and print the number of unique values in the `Country` column.

```python
# Sort by item_price
print(data.sort_values(by="item_price").head())

# Sort by order_quantity descending
print(data.sort_values(by="order_quantity", ascending=False).head())

# Number of unique item names
print(data["item_name"].nunique())

# All unique item names
print(data["item_name"].unique())

# Number of unique countries
print(data["Country"].nunique())
```
